## 2. Displaying Information (4 minutes)

**Concept:** `print()` displays information. Text written between matching single or double quotation marks is a **string literal**. The quotation marks show Python where the text begins and ends; they are not included in the displayed message.

**Example and predict:** Read the message and predict what will appear. Then run the cell.

````python
print("I am learning Python to solve problems.")
````

**Example:** Each call to `print()` displays its message on a new line. A line beginning with `#` is a **comment**: Python treats it as a note for people reading the code.

**Predict, then run:** How many lines of output will you see?

````python
# These messages are about learning and software development.
print("Practise one step at a time.")
print("Test your code as you go.")
````

## 3. Python as a Calculator (4 minutes)

**Concept:** Python can calculate with the arithmetic operators `+` (add), `-` (subtract), `*` (multiply), and `/` (divide). You can write an expression directly in a code cell, or ask `print()` to display it.

**Example:** Imagine you spend 2 hours revising and 1 hour practising code. Predict each result, then run the examples.

````python
print("Revision and practice:", 2 + 1, "hours")
print("Time left from a 7-hour study plan:", 7 - 2, "hours")
print("Two practice sessions of 3 hours:", 2 * 3, "hours")
print("Half of a 6-hour study plan:", 6 / 2, "hours")
````

**Predict Before You Run:** If you study for 3 hours each week for 7 weeks, how many learning hours is that altogether? Write down your prediction, then run the calculation.

**Experiment:** Change one of the numbers to match your own study plan and run it again. **Reflect:** Does the result make sense for the number of weeks?

````python
print("Learning hours over seven weeks:", 3 * 7)
````

## 4. Getting User Input (5 minutes)

**Concept:** `input()` displays a prompt and waits for someone to type a response. It returns the response as text. In this example, `name` stores the entered value so that the next line can use it; we will study names that store values in more detail in Session 2.

**Predict:** What will the prompt say, and what will the welcome message include? Then run the cell and enter your name.

````python
name = input("Enter your name: ")
print("Welcome,", name)
````

**Example:** To use a typed number in a calculation, convert the numeric text with `int()`. Here, the prompt asks how many hours you study each week, and the programme multiplies that number by seven.

**Predict, then run:** Predict the total for 3 hours per week. Enter a whole number when prompted. (`int()` is a practical conversion for this calculation; we will keep the explanation simple for now.)

````python
hours_per_week = int(input("How many hours will you study each week? "))
total_hours = hours_per_week * 7
print("Your learning hours over seven weeks:", total_hours)
````
