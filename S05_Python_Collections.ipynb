## 1. Why Collections Matter (2 minutes)

**Concept**

Imagine a learner completes four study sessions: Python for 45 minutes, Mathematics for 30 minutes, Python for 60 minutes, and English for 25 minutes. One way to keep the durations is to give each one a separate variable:

````python
first_duration = 45
second_duration = 30
third_duration = 60
fourth_duration = 25

print(first_duration, second_duration, third_duration, fourth_duration)
````

This works for four sessions, but managing many separate names becomes inconvenient. A **collection** keeps related values together. A list is one collection that can hold the session durations in their recording order.

**Predict:** What will `len(study_durations)` and `sum(study_durations)` return?

````python
study_durations = [45, 30, 60, 25]

print(len(study_durations))
print(sum(study_durations))
````

## 2. Lists: Creating, Accessing and Updating (5 minutes)

**Concept**

A list preserves element order, allows duplicate values, and is **mutable**, which means its contents can be changed. Each item has an index: the first index is `0`, and a negative index counts from the end.

**Predict:** What will each of the three print statements display?

````python
study_durations = [45, 30, 60, 25]

print(study_durations[0])
print(study_durations[-1])
print(len(study_durations))
````

**Run**

Accessing an item reads its value. Assigning a value at an index modifies the list at that position. Here, the second session duration changes from 30 to 40 minutes.

````python
study_durations[1] = 40
print(study_durations)
````

## 3. Lists: Iteration, Slicing and Common Methods (4 minutes)

**Concept**

A `for` loop can visit each value in a list. Slicing selects a range of values: the starting index is included and the ending index is excluded.

**Predict:** What will each slice contain?

````python
study_durations = [45, 30, 60, 25]

for duration in study_durations:
    print(duration)

print(study_durations[1:3])
print(study_durations[:2])
print(study_durations[-2:])
````

**Run**

Common methods change a list. `append()` adds one object as a single item. `extend()` adds individual items from another iterable. `remove()` removes the first matching value, while `pop()` removes and returns an item (the final item by default).

````python
durations = [30, 45]
durations.append(30)
durations.extend([60, 25])
durations.remove(45)
last_duration = durations.pop()

print(durations)
print(last_duration)
````

`append([30, 45])` would add the entire list as one nested item; `extend([30, 45])` adds `30` and `45` as separate items. In the example above, `append(30)` adds one duration, and `extend([60, 25])` adds two durations.

````python
practice_durations = [20, 35, 50]
practice_durations.append(40)
practice_durations.remove(35)

print(practice_durations)
print(sum(practice_durations))
````

## 4. Tuples: Fixed Data and Immutability (3 minutes)

**Concept**

A tuple is an ordered sequence, like a list, but its elements cannot be reassigned. It is useful for a group of values intended to stay fixed. The weekly goal below stores a target number of sessions and a target number of study minutes.

**Predict:** What will `len(weekly_goal)` return?

````python
weekly_goal = (5, 180)

print(weekly_goal[0])
print(weekly_goal[1])
print(len(weekly_goal))
````

The tuple's indexes work like list indexes, but an assignment such as `weekly_goal[0] = 6` is invalid. If run, it raises `TypeError` because tuple elements cannot be reassigned. This example is intentionally shown as text, not executed.

A tuple being immutable does not make a mutable object stored inside it immutable. For this session, use tuples for simple fixed values and avoid more complex nested collections.

## 5. Sets: Uniqueness and Membership (4 minutes)

**Concept**

A set keeps unique values, so repeated topic names are represented once. Sets do not support positional indexing and do not guarantee iteration order. They are useful for efficient membership checks. Use `sorted()` when you need a predictable display order.

**Predict:** How many unique topics will there be? Will `"Python"` be a member? What order will the sorted result use?

````python
topics = ["Python", "English", "Python", "Mathematics", "English"]
unique_topics = set(topics)

print(len(unique_topics))
print("Python" in unique_topics)
print(sorted(unique_topics))
````

**Predict → Run → Reflect**

Compare the list and set below. Predict the length of each before running the cell. The list preserves its repeated entries; the set contains each topic only once. Its sorted view makes the display order predictable.

````python
repeated_topics = ["Python", "English", "Python", "Mathematics", "English"]
distinct_topics = set(repeated_topics)

print(len(repeated_topics), repeated_topics)
print(len(distinct_topics), sorted(distinct_topics))
````

**Run**

`add()` inserts a value. `remove()` deletes a value and expects it to be present; `discard()` is a safer choice when a value may be absent. Union (`|`) combines values from both sets, and intersection (`&`) keeps values present in both.

````python
morning_topics = {"Python", "Mathematics"}
afternoon_topics = {"English", "Python"}

morning_topics.add("History")
morning_topics.remove("History")
morning_topics.discard("Art")

print(sorted(morning_topics | afternoon_topics))
print(sorted(morning_topics & afternoon_topics))
````

**Experiment:** Add another repeated topic to `repeated_topics` and run the comparison again. What changes in the list, and what changes in the set?

**Reflect:** Why is a set a better choice than a list when we only need unique topic names?

## 6. Dictionaries: Keys, Values and Accumulation (5 minutes)

**Concept**

A dictionary associates a **key** with a **value**. Keys identify entries and must be unique; values may repeat. Modern Python dictionaries preserve insertion order. Assigning to an existing key replaces its value.

This dictionary records total study minutes by topic. **Predict:** What values will the two lookups display?

````python
study_time = {
    "Python": 105,
    "Mathematics": 30,
    "English": 25
}

print(study_time["Python"])
print(study_time.get("History", 0))
````

`get(key, default)` returns the default value when a key is missing, so it can prevent a `KeyError`. A direct lookup such as `study_time["Python"]` is appropriate when you know that key exists.

**Run:** Assigning a new key adds an entry; assigning an existing key updates it. When repeated sessions should be totalled, accumulate the new duration rather than replacing the old value.

````python
study_time["History"] = 35
study_time["English"] = 20
study_time["Python"] = study_time.get("Python", 0) + 45

print(study_time)
````

**Predict:** Before running the next cell, what will each pair printed by `.items()` contain? Dictionaries also provide `.keys()` for keys and `.values()` for values.

````python
for topic, minutes in study_time.items():
    print(topic, minutes)

print(list(study_time.keys()))
print(list(study_time.values()))
print(list(study_time.items()))
````

## 8. Debugging AI-Generated Collection Code (3 minutes)

Code can be syntactically valid and still have a logic defect. For each investigation, follow **Predict → Identify the defect → Correct → Verify**. Check what the program does, not only whether it runs.

### Bug A: Incorrect dictionary accumulation

**Predict:** What dictionary will this loop produce? Will it contain the total of both Python sessions?

**Identify the defect:** Each assignment replaces the value already stored for that topic.

````python
study_time = {}

for topic, duration in [("Python", 30), ("Python", 45)]:
    study_time[topic] = duration

print(study_time)
````

**Correct → Verify:** Use the previous value (or zero if absent) and add the new duration. Predict the corrected total before running the verification.

````python
study_time = {}

for topic, duration in [("Python", 30), ("Python", 45)]:
    study_time[topic] = study_time.get(topic, 0) + duration

print(study_time)
````

### Bug B: Unexpected list mutation

**Predict:** Will `original` change after an item is appended to `updated`? Explain your prediction before running the code.

````python
original = [30, 45, 60]
updated = original
updated.append(20)

print(original)
````

**Identify the defect:** Assignment did not make a separate list. Both names refer to the same mutable list, so changing it through either name is visible through the other.

**Correct → Verify:** `.copy()` makes an independent list for this simple example.

````python
original = [30, 45, 60]
updated = original.copy()
updated.append(20)

print(original)
print(updated)
````

### Bug C: Unsafe dictionary lookup

**Predict → Identify the defect:** The dictionary below has no `"History"` key. What exception would a direct lookup raise? This faulty lookup is Markdown only and will not interrupt **Run all**.

```python
study_time = {"Python": 60}
print(study_time["History"])
```

**Correct → Verify:** Use `get()` with an appropriate default when the key may be absent.

````python
study_time = {"Python": 60}
print(study_time.get("History", 0))
````
