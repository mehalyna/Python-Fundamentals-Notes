## 1. From Loops to Comprehensions — 3 minutes

**Concept.** A list comprehension is a compact way to construct a list. It follows the same logic as a familiar loop, but shorter code is not automatically clearer.

**Predict.** What happens to a duration of exactly 30 minutes?

````python
durations = [45, 30, 60, 25]

focused_with_loop = []
for duration in durations:
    if duration >= 30:
        focused_with_loop.append(duration)

focused_with_comprehension = [
    duration
    for duration in durations
    if duration >= 30
]

print("Loop:", focused_with_loop)
print("Comprehension:", focused_with_comprehension)
print("Same result:", focused_with_loop == focused_with_comprehension)
````

**Experiment.** Change both `>= 30` conditions to `> 30`. Which value disappears?

**Reflect.** Why is the inclusive operator `>=` important when the requirement says “at least 30 minutes”?

## 2. List Comprehensions for Transformation — 4 minutes

**Concept.** The basic form is `new_list = [expression for item in collection]`. Each input value is transformed into one output value.

**Predict.** What will the first item in `hours` be? Will `durations` change?

````python
durations = [45, 30, 60, 25, 45]

hours = [
    round(duration / 60, 2)
    for duration in durations
]
extended_durations = [duration + 5 for duration in durations]

print("Hours:", hours)
print("With five extra minutes:", extended_durations)
print("Original durations:", durations)
````

**Reflect.** Does the comprehension modify the original list? Why is creating a new list useful for the Study Insights Generator?

## 3. List Comprehensions for Filtering — 4 minutes

## 4. Combining Transformation and Filtering — 4 minutes

**Concept.** A comprehension can filter first and transform only qualifying values. The logical sequence is: iterate, test, transform, collect.

**Predict.** With a threshold of 30, will the duration exactly equal to 30 be included?

````python
durations = [45, 30, 60, 25, 45]
minimum_duration = 30

focused_hours = [
    round(duration / 60, 2)
    for duration in durations
    if duration >= minimum_duration
]
focused_hours_with_loop = []
for duration in durations:
    if duration >= minimum_duration:
        focused_hours_with_loop.append(round(duration / 60, 2))

print("Comprehension:", focused_hours)
print("Equivalent loop:", focused_hours_with_loop)
print("Same result:", focused_hours == focused_hours_with_loop)
````

**Engineering Check.** Is combining the transformation and filter readable here? Avoid nested comprehensions when they hide the logic.

### Nested Lists: Working with Simple Matrices

**Concept.** A two-dimensional list is a list whose items are lists. It is a beginner-friendly way to represent a rectangular table: each inner list is a row, and each item in a row is a column. The first index selects a row; the second index selects a column, so `matrix[row][column]` means “go to this row, then this column”.

We will use a 3×3 study-hours matrix. Rows represent students and columns represent Monday, Tuesday and Wednesday. Python counts both row and column indices from zero.

**Predict 1.** What value is at row `1`, column `2`? Which student does row `0` represent?

````python
study_hours = [
    [2, 1, 3],  # Alex: Monday, Tuesday, Wednesday
    [1, 2, 2],  # Bea
    [3, 2, 1],  # Chandra
]

print("Bea on Wednesday:", study_hours[1][2])
print("Alex's complete row:", study_hours[0])
print("Number of rows:", len(study_hours))
print("Number of columns:", len(study_hours[0]))
````

**Reflect.** `study_hours[1]` is a complete row, while `study_hours[1][2]` is one cell. A row index and a column index are both positions, but they describe different directions in the table.

**Experiment 1.** Change Bea's Wednesday value and retrieve Chandra's complete row. Then try `study_hours[3][0]` in a separate cell. What boundary error do you get?

````python
study_hours = [
    [2, 1, 3],
    [1, 2, 2],
    [3, 2, 1],
]

study_hours[1][2] = 4
print("Updated matrix:", study_hours)
print("Chandra's row:", study_hours[2])
print("Rows versus columns:", len(study_hours), len(study_hours[0]))
````

**Concept.** Nested `for` loops visit each row and then each value inside that row. This is useful when the calculation or display rule needs more than a short expression.

**Predict 2.** How many numbers will be printed? Which loop variable represents a row, and which represents one study day?

**Experiment 2.** Change the labels or add a fourth day to every row. **Edge case:** rectangular matrices are easier to process because every row has the same number of columns.

````python
study_hours = [
    [2, 1, 3],
    [1, 2, 2],
    [3, 2, 1],
]
students = ["Alex", "Bea", "Chandra"]
days = ["Monday", "Tuesday", "Wednesday"]

for row_index, row in enumerate(study_hours):
    print(students[row_index])
    for column_index, hours in enumerate(row):
        print(" ", days[column_index], hours, "hours")
````

````python
study_hours = [
    [2, 1, 3],
    [1, 2, 2],
    [3, 2, 1],
]

row_sums = []
for row in study_hours:
    row_sums.append(sum(row))

column_sums = []
for column_index in range(len(study_hours[0])):
    column_total = 0
    for row in study_hours:
        column_total += row[column_index]
    column_sums.append(column_total)

total_hours = sum(row_sums)
print("Row sums:", row_sums)
print("Column sums:", column_sums)
print("Total hours:", total_hours)
````

**Reflect.** Row sums answer “how much did each student study?”. Column sums answer “how many hours did the group study each day?”. The grand total can be calculated from either set of sums, but both should agree.

**Engineering check.** `len(matrix)` counts rows. `len(matrix[0])` counts columns only when the matrix is non-empty and its first row exists. Validate those assumptions before indexing external data.

````python
rows = 3
columns = 4
constructed = []
for row_number in range(rows):
    row = []
    for column_number in range(columns):
        row.append(row_number + column_number)
    constructed.append(row)

print("Constructed matrix:", constructed)
print("Dimensions:", len(constructed), "rows x", len(constructed[0]), "columns")
````

### Debugging example: repeated row aliases

**Concept.** `[[0] * 3] * 3` repeats references to one inner list. The code runs, but changing one row changes every row. A list comprehension creates a new inner list on each repetition.

**Predict.** After changing `bad_matrix[0][1]`, how many rows will contain `9`?

````python
bad_matrix = [[0] * 3] * 3
bad_matrix[0][1] = 9
print("Incorrect aliased matrix:", bad_matrix)

good_matrix = [[0] * 3 for _ in range(3)]
good_matrix[0][1] = 9
print("Correct independent rows:", good_matrix)
````

**Expected:** the incorrect matrix has `9` in every row; the corrected matrix has `9` only in its first row.

### Transposing a matrix

**Concept.** A transpose turns rows into columns. `zip(*matrix)` uses `*` to unpack the outer list, passing each row as a separate argument to `zip`. The result is an iterator of tuples, so convert each tuple to a list when a list-of-lists result is wanted.

**Predict 3.** What will the first row of the transposed matrix contain?

````python
study_hours = [
    [2, 1, 3],
    [1, 2, 2],
    [3, 2, 1],
]

transposed = [list(column) for column in zip(*study_hours)]
print("Original:", study_hours)
print("Transposed:", transposed)
print("Monday column is now:", transposed[0])
````

**Edge cases.** An empty matrix has no first row, so `len(matrix[0])` raises `IndexError`. Rows of different lengths are a jagged matrix: some column positions exist in one row but not another. Rectangular matrices make nested loops, column sums and transposition predictable.

**Experiment 4.** Try the two boundary examples below. Decide whether your real study-data program should reject jagged input, pad missing cells or report only complete rows.

## 12. Sets: Methods and Algebra — Core exploration

**Concept.** Sets store unique hashable values. `update`, `add`, `discard`, `remove`, `pop` and `clear` manage membership. Operators and methods express union, intersection, difference, symmetric difference, subset and superset relationships.

**Predict.** Which duplicate topics disappear? Which topics are in both study groups?

````python
python_topics = {"lists", "loops", "dictionaries"}
maths_topics = {"lists", "fractions"}
print("Union:", sorted(python_topics | maths_topics))
print("Intersection:", python_topics & maths_topics)
print("Difference:", python_topics - maths_topics)
print("Symmetric difference:", sorted(python_topics ^ maths_topics))
````

````python
python_topics = {"lists", "loops", "dictionaries"}
maths_topics = {"lists", "fractions"}
print("Method union:", python_topics.union(maths_topics))
print("Method intersection:", python_topics.intersection(maths_topics))
print("Method difference:", python_topics.difference(maths_topics))
print("Method symmetric difference:", python_topics.symmetric_difference(maths_topics))
````

````python
completed = {"lists", "loops"}
all_python = {"lists", "loops", "dictionaries"}
print("Subset operator:", completed <= all_python)
print("Subset method:", completed.issubset(all_python))
print("Superset operator:", all_python >= completed)
print("Superset method:", all_python.issuperset(completed))
removed = all_python.pop()
print("Arbitrary pop:", removed, "remaining:", all_python)
all_python.clear()
print("Cleared:", all_python)
````

## 13. Dictionaries: Methods, Rules and Merging — Core exploration

**Concept.** Dictionary keys are unique and hashable; values may be mutable. Assignment overwrites a key, while explicit accumulation adds to its existing value. `get` reads safely; `setdefault` inserts a missing default. `keys`, `values` and `items` support iteration.

**Predict.** What happens when `Python` is assigned twice? Does `get` add a missing key?

**Experiment.** Merge dictionaries with overlapping keys in both orders. **Edge case:** dictionary insertion order is preserved, but merging is not aggregation; the right-hand value wins for a duplicate key.

````python
study_time = {"Python": 105, "Maths": 30}
print("Read:", study_time["Python"])
study_time["English"] = 25
study_time["Python"] = 110
study_time.update({"Maths": 35, "Science": 20})
print("Updated:", study_time)
print("Keys:", list(study_time.keys()))
print("Values:", list(study_time.values()))
print("Items:", list(study_time.items()))
````

````python
study_time = {"Python": 105}
print("Missing without mutation:", study_time.get("History", 0), study_time)
print("Existing value:", study_time.get("Python", 0))
new_value = study_time.setdefault("History", 0)
study_time.setdefault("Python", 999)
print("Inserted default:", new_value)
print("After setdefault:", study_time)
````

````python
progress = {"Python": 105, "Maths": 30, "English": 25}
removed = progress.pop("Maths")
missing = progress.pop("History", 0)
print("Removed:", removed, "safe missing:", missing)
last_pair = progress.popitem()
print("Last pair:", last_pair)
progress.clear()
print("Cleared:", progress)
````

````python
morning = {"Python": 45, "Maths": 30}
evening = {"Maths": 20, "English": 25}
print("Morning then evening:", morning | evening)
print("Evening then morning:", evening | morning)
morning |= evening
print("In-place merge:", morning)
totals = {}
for topic, minutes in [("Python", 25), ("Python", 20)]:
    totals[topic] = totals.get(topic, 0) + minutes
print("Accumulation, not merge:", totals)
````

**Engineering check.** State whether duplicate keys overwrite or accumulate before choosing assignment, `|`, or an explicit total.
