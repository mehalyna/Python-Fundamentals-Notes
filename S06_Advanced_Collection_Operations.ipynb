## 1. From Loops to Comprehensions — 3 minutes

**Concept.** A list comprehension is a compact way to construct a list. It follows the same logic as a familiar loop, but shorter code is not automatically clearer.

**Predict.** What happens to a duration of exactly 30 minutes?

````python
durations = [45, 30, 60, 25]

focused_with_loop = []
for duration in durations:
    if duration >= 30:
        focused_with_loop.append(duration)

focused_with_comprehension = [
    duration
    for duration in durations
    if duration >= 30
]

print("Loop:", focused_with_loop)
print("Comprehension:", focused_with_comprehension)
print("Same result:", focused_with_loop == focused_with_comprehension)
````

**Experiment.** Change both `>= 30` conditions to `> 30`. Which value disappears?

**Reflect.** Why is the inclusive operator `>=` important when the requirement says “at least 30 minutes”?

## 2. List Comprehensions for Transformation — 4 minutes

**Concept.** The basic form is `new_list = [expression for item in collection]`. Each input value is transformed into one output value.

**Predict.** What will the first item in `hours` be? Will `durations` change?

````python
durations = [45, 30, 60, 25, 45]

hours = [
    round(duration / 60, 2)
    for duration in durations
]
extended_durations = [duration + 5 for duration in durations]

print("Hours:", hours)
print("With five extra minutes:", extended_durations)
print("Original durations:", durations)
````

**Experiment.** Change the rule to add 10 minutes, or convert minutes to hours without rounding. How does the output change?

**Reflect.** Does the comprehension modify the original list? Why is creating a new list useful for the Study Insights Generator?

## 4. Combining Transformation and Filtering — 4 minutes

**Concept.** A comprehension can filter first and transform only qualifying values. The logical sequence is: iterate, test, transform, collect.

**Predict.** With a threshold of 30, will the duration exactly equal to 30 be included?

````python
durations = [45, 30, 60, 25, 45]
minimum_duration = 30
+
+focused_hours = [
+    round(duration / 60, 2)
+    for duration in durations
+    if duration >= minimum_duration
+]
+
+focused_hours_with_loop = []
+for duration in durations:
+    if duration >= minimum_duration:
+        focused_hours_with_loop.append(round(duration / 60, 2))
+
+print("Comprehension:", focused_hours)
+print("Equivalent loop:", focused_hours_with_loop)
+print("Same result:", focused_hours == focused_hours_with_loop)
````

**Experiment.** Change `minimum_duration` to 45, then to 61. Predict the list before running each version.

+**Engineering Check.** Is combining the transformation and filter readable here? Avoid nested comprehensions when they hide the logic.

## 5. Dictionary Comprehensions: Transforming Values — 4 minutes

**Concept.** The form is `new_dict = {key_expression: value_expression for key, value in dictionary.items()}`. `.items()` provides each key and value, which are unpacked into `topic` and `minutes`.

**Predict.** What values will Python and Mathematics have in hours?

````python
study_time = {
    "Python": 105,
    "Mathematics": 30,
    "English": 25,
    "Data Science": 45
}

topic_hours = {
    topic: round(minutes / 60, 2)
    for topic, minutes in study_time.items()
+    }
+
+print("Topic hours:", topic_hours)
+print("Original minutes:", study_time)
````

## 6. Dictionary Comprehensions: Filtering Entries — 4 minutes

**Concept.** Filter dictionary entries by putting a condition after `.items()`. Here the condition applies to the accumulated value, not the topic name.

**Predict.** Which topics qualify when the threshold is 40 minutes?

````python
study_time = {
    "Python": 105,
    "Mathematics": 30,
    "English": 25,
    "Data Science": 45
}
minimum_minutes = 40
+
+qualifying_topics = {
+    topic: minutes
+    for topic, minutes in study_time.items()
+    if minutes >= minimum_minutes
+}
+
+print(qualifying_topics)
+print("Alphabetical display:")
+for topic in sorted(qualifying_topics):
+    print(topic, qualifying_topics[topic])
````

**Experiment.** Set the threshold to `0`, then to `200`. Predict the qualifying dictionary before running each version.
+
+**Engineering Check.** The comprehension creates a new dictionary. `sorted()` is used only for predictable alphabetical display; it does not change the dictionary.

## 7. Choosing Between Loops and Comprehensions — 2 minutes

**Engineering Check.** Choose the clearest implementation, not automatically the shortest one.
+
+1. `hours = [round(minutes / 60, 2) for minutes in durations]` is a simple transformation suited to a comprehension.
+2. `long_sessions = [duration for duration in durations if duration >= 45]` is a simple filter suited to a comprehension.
+3. Validation, dictionary accumulation and several side effects are clearer as a regular loop:
+
+```python
+topic_totals = {}
+for topic, minutes in activities:
+    if minutes > 0:
+        if topic not in topic_totals:
+            topic_totals[topic] = 0
+        topic_totals[topic] += minutes
+```
+
+Comprehensions construct collections. Ordinary loops remain appropriate for multi-step processing. Avoid comprehensions used only for side effects.

````python
durations = [45, 30, 60, 25, 45]
+    
+hours = [round(minutes / 60, 2) for minutes in durations]
+long_sessions = [duration for duration in durations if duration >= 45]
+
+print(hours)
+print(long_sessions)
````
