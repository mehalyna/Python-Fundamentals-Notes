## 2. Comparison Operators and Boolean Expressions (4 minutes)

A comparison asks a question about values. Its result is a Boolean value: `True` or `False`.

| Operator | Meaning | Example question |
| --- | --- | --- |
| `==` | equal to | Is the score equal to 70? |
| `!=` | not equal to | Is the score different from 70? |
| `>` | greater than | Is the score above 70? |
| `<` | less than | Is the score below 70? |
| `>=` | greater than or equal to | Is the score at least 70? |
| `<=` | less than or equal to | Is the score no more than 70? |

**Concept:** `=` assigns a value to a variable; `==` compares two values. They have different jobs.

````python
hours = 4.5

print(hours >= 3)
print(hours < 6)
print(hours == 4.5)
````

**Predict Before Running:** The academy's Balanced Study Plan starts at 3 hours. What do you expect each comparison to return for 2.9, 3 and 3.1 hours? Predict whether each value is at least 3, then run the cell.

````python
hours = 2.9
print(hours >= 3)

hours = 3
print(hours >= 3)

hours = 3.1
print(hours >= 3)
````

**Reflect:** Which single comparison changes from `False` to `True` at the threshold? A reliable decision rule must handle the value exactly on the boundary.

## 3. Making Decisions with `if` and `else` (4 minutes)

An `if` statement runs its indented block when its condition is `True`. `else` runs when that condition is `False`.

**Run:** The colon starts the block. Indentation (four spaces here) marks which instructions belong to it; indentation is part of Python's syntax.

````python
hours = 4

if hours >= 3:
    print("You have enough time for a regular study plan.")
else:
    print("Consider a lighter study plan.")
````

## 4. Multiple Decisions with `elif` (4 minutes)

Assessment feedback has three categories: below 40 is **Further Practice Needed**; 40 to less than 75 is **Developing Well**; 75 or above is **Strong Progress**.

`elif` means "else if". Python checks an `if` / `elif` / `else` chain from top to bottom and runs only the **first** branch whose condition is true. This makes the order important.

````python
score = 62

if score < 40:
    print("Further Practice Needed")
elif score < 75:
    print("Developing Well")
else:
    print("Strong Progress")
````

**Predict, then Try It:** Change `score` to **39**, **40**, **74** and **75**, running the cell each time. Check your results against these categories: 39 -> Further Practice Needed; 40 and 74 -> Developing Well; 75 -> Strong Progress. The ranges do not overlap because the earlier branch takes priority.

````python
score = 39  # Try 40, 74 and 75 as well.

if score < 40:
    print("Further Practice Needed")
elif score < 75:
    print("Developing Well")
else:
    print("Strong Progress")
````

## 5. Logical Operators: `and`, `or`, `not` (5 minutes)

Logical operators combine or change Boolean values:

- `and`: both conditions must be true.
- `or`: at least one condition must be true.
- `not`: reverses a Boolean value.

A student can access an optional mentoring session if they have completed the introductory exercises **and** have at least two hours available, **or** if they have a mentor invitation. Parentheses show that the two requirements in the first route belong together.

````python
completed_intro = True
hours_available = 2
has_invitation = False

can_attend = (completed_intro and hours_available >= 2) or has_invitation
print(can_attend)
````

**Concept:** A truth table shows how `and` and `or` combine two Boolean values.

| First value | Second value | `First and Second` | `First or Second` |
| --- | --- | --- | --- |
| `False` | `False` | `False` | `False` |
| `False` | `True` | `False` | `True` |
| `True` | `False` | `False` | `True` |
| `True` | `True` | `True` | `True` |

Parentheses make complex conditions easier to read and help communicate the intended grouping.

````python
first_condition = False
second_condition = True

print(first_condition and second_condition)
print(first_condition or second_condition)
print(not first_condition)
````

**Predict, then Experiment:** Predict the three results first. Change one Boolean value at a time and run again. For the mentoring rule, what happens if the student has an invitation but has not completed the exercises? What if they have completed the exercises but have only one hour and no invitation?

````python
completed_intro = False  # Try True.
hours_available = 1  # Try 2.
has_invitation = False  # Try True.

can_attend = (completed_intro and hours_available >= 2) or has_invitation
print(can_attend)
````

## 6. Nested Conditions and Readability (3 minutes)

A nested `if` is an `if` inside another `if`. Here, a student must pass an initial assessment; only then do we check whether the practical exercises are complete.

````python
passed_assessment = True
completed_practical = False

if passed_assessment:
    if completed_practical:
        print("You have completed this stage.")
    else:
        print("Complete the practical exercises to finish this stage.")
else:
    print("Review your assessment feedback and try again.")
````

When both checks lead to the same outcome, they can sometimes be combined with `and`. Keep nesting when it makes the steps clearer; avoid making conditions hard to read.

````python
passed_assessment = True
completed_practical = True

if passed_assessment and completed_practical:
    print("You have completed this stage.")
else:
    print("There is still a stage requirement to complete.")
````

## Additional Section. Decision Making with `match`/`case` (4 minutes)

`match`/`case` was introduced in Python 3.10. It is similar to `switch`/`case` in some other programming languages, but Python can support more advanced structural pattern matching. For now, we will use only simple constant values and the wildcard `_` default case.

**Concept:** `match` selects a value to inspect. Each `case` defines a pattern to match. Python runs only the first matching case, and `case _` handles values that do not match earlier cases. Python does not require `break` statements between cases.

This example recommends an approach for a known learning mode:

````python
learning_mode = "practical"

match learning_mode:
    case "practical":
        print("Focus on coding exercises.")
    case "theory":
        print("Start with guided explanations.")
    case "mixed":
        print("Combine explanations and practice.")
    case _:
        print("Unknown learning mode.")
````

**Predict -> Run -> Experiment:** Before running, predict the output for each of these values: `"practical"`, `"theory"`, `"mixed"` and `"other"`. Change `learning_mode` one value at a time and run the cell. The last value should use the wildcard default case.

````python
learning_mode = "theory"  # Try "practical", "mixed" and "other".

match learning_mode:
    case "practical":
        print("Focus on coding exercises.")
    case "theory":
        print("Start with guided explanations.")
    case "mixed":
        print("Combine explanations and practice.")
    case _:
        print("Unknown learning mode.")
````

**Compare:** Use `if`/`elif`/`else` for ranges, comparisons and complex Boolean conditions, such as `score >= 70 and practical_complete`. Use simple `match`/`case` when selecting behaviour based on distinct, known values such as learning modes. Neither form is always better or faster; choose the one that expresses the decision rule clearly.

`match`/`case` requires Python 3.10 or later. This notebook does not use advanced patterns such as guards, sequence matching, mapping matching or class patterns.
