## 1. Why Do We Need Modules? (3 minutes)

**CONCEPT:** A module is a Python file containing reusable code. Keeping every function in one large file can become difficult to read and maintain. Moving calculations to `calculations.py` changes their location, not their expected behaviour.

```text
main.py  ──imports──>  calculations.py
                         calculate_total()
                         filter_sessions()
                         convert_to_hours()
```

**PREDICT:** What should change when these functions move to another file? What should stay the same?

````python
def calculate_total(durations):
    return sum(durations)

def filter_sessions(durations, minimum_duration):
    return [
        duration
        for duration in durations
        if duration >= minimum_duration
    ]

def convert_to_hours(durations):
    return [round(duration / 60, 2) for duration in durations]

durations = [45, 30, 60, 25]
print(calculate_total(durations))
print(filter_sessions(durations, 30))
print(convert_to_hours(durations))
````

**EXPERIMENT:** Imagine adding ten more Learning Assistant functions. Which functions would you group in a calculations module?

**REFLECT:** What is the benefit of giving related reusable code its own file?

**Expected result:** The function outputs do not depend on which file contains the definitions.

## 2. Importing Standard-Library Modules (4 minutes)

**CONCEPT:** Python's standard library is a collection of modules that are included with Python. `import math` gives access to the module namespace; the dot connects the module name to one of its functions. `from math import sqrt` imports one name directly. An alias gives a module a shorter local name.

**PREDICT:** What are the results of `sqrt(81)`, `math.ceil(2.3)` and `m.sqrt(49)`?

````python
import math
from math import sqrt
import math as m

print(math.sqrt(81))
print(sqrt(81))
print(math.ceil(2.3), math.floor(2.9))
print(m.sqrt(49))
assert math.sqrt(81) == sqrt(81) == 9.0
assert m.sqrt(49) == 7.0
````

**EXPERIMENT:** Import `math` with a different clear alias, such as `math_tools`, and call `math_tools.floor(4.8)`. Try to keep aliases readable rather than mysterious.

**REFLECT:** Which is clearer in a large programme: `math.sqrt(value)` or a short alias such as `m.sqrt(value)`? Why?

**Expected result:** `math.sqrt(81)` and `sqrt(81)` are equivalent only because both imports were made; `m.sqrt(49)` works through the alias.

## 3. Useful Standard-Library Modules (4 minutes)

**CONCEPT:** `math` supports numerical work, `random` supports choices, and `datetime` represents dates and time. Random choices and today's date can vary between runs, so fixed dates are useful for predictable teaching examples.

**PREDICT:** What date is seven days after 28 October 2026? Why should we avoid adding seven to a date string ourselves?

````python
import random
from datetime import date, timedelta

random.seed(7)
print(random.choice(["Python", "Mathematics", "English"]))

today = date.today()
print("Runtime date:", today)

start = date(2026, 10, 28)
deadline = start + timedelta(days=7)
assert deadline.isoformat() == "2026-11-04"
study_deadline = start + timedelta(days=14)
print("Fixed deadline:", deadline)
print("Fourteen-day study deadline:", study_deadline)
````

**EXPERIMENT:** Change `start` to a date near the end of a month and calculate a seven-day deadline.

**REFLECT:** Why is `timedelta` safer than manually changing the day number? It handles different month lengths and leap years for us.

**Expected result:** The fixed seven-day deadline is 4 November 2026, and the fourteen-day deadline is 11 November 2026.

## 4. Creating Your Own Module in Google Colab (4 minutes)

**CONCEPT:** A `.py` file is a real Python module. In Colab, `%%writefile` writes the cell contents to the current runtime. It must be the first line of its cell.

**PREDICT:** After importing `calculations`, what namespace will contain `calculate_total`?

````python
%%writefile calculations.py
def calculate_total(durations):
    return sum(durations)

def filter_sessions(durations, minimum_duration):
    return [
        duration
        for duration in durations
        if duration >= minimum_duration
    ]
````

````python
import calculations
durations = [45, 30, 60, 25]
print(calculations.calculate_total(durations))
print(calculations.filter_sessions(durations, 30))
assert calculations.calculate_total(durations) == 160
assert calculations.filter_sessions(durations, 30) == [45, 30, 60]
````

**EXPERIMENT:** Add a `convert_to_hours()` function to the `%%writefile` cell, rerun that cell, then run the reload-and-check cell below. Python caches imported modules, so editing a file does not automatically update an existing import.

**Expected result:** `importlib.reload(calculations)` reads the current file again. The module name is used before its functions because the functions live in that namespace.

````python
import importlib
importlib.reload(calculations)
assert calculations.calculate_total([30, 60]) == 90
print("Module reloaded successfully.")
````

## 5. From Modules to Packages (4 minutes)

**CONCEPT:** A package organises related modules under a common name. Here, `study_toolkit` is a folder, `__init__.py` marks its initialisation, and `calculations.py` contains calculation functions.

```text
study_toolkit/
├── __init__.py
└── calculations.py
```

The import `from study_toolkit.calculations import calculate_total` means: enter package `study_toolkit`, find module `calculations`, then import its function `calculate_total`.

**PREDICT:** Is a package the same thing as one module?

````python
from pathlib import Path

Path("study_toolkit").mkdir(exist_ok=True)
````

````python
%%writefile study_toolkit/__init__.py
# Study Toolkit package
````

````python
%%writefile study_toolkit/calculations.py
def calculate_total(durations):
    return sum(durations)

def convert_to_hours(durations):
    return [
        round(duration / 60, 2)
        for duration in durations
    ]
````

````python
from study_toolkit.calculations import calculate_total, convert_to_hours
durations = [30, 60, 90]
print(calculate_total(durations))
print(convert_to_hours(durations))
assert calculate_total(durations) == 180
assert convert_to_hours(durations) == [0.5, 1.0, 1.5]
````

**EXPERIMENT:** Create a second module named `reports.py` containing `format_total(minutes)`, then import it with `from study_toolkit.reports import format_total`. Remember to create the directory before writing the file.

**REFLECT:** A module is one Python file; a package is a named folder containing related modules. Modern Python also supports namespace packages without `__init__.py`, but we use the conventional file here to make the package explicit.

**Expected result:** The package functions return the same values as before, even though their path is now longer.

## 6. PIP and Third-Party Libraries (3 minutes)

**CONCEPT:** Built-in functions such as `sum()` are always available. Standard-library modules such as `datetime` come with Python. Third-party packages are distributed separately. PIP is Python's package installer; `%pip` is the notebook-friendly command.

**PREDICT:** Should we install an external package just to display today's date?

````python
import sys
from importlib import metadata
print("Python:", sys.version.split()[0])
try:
    print("PIP:", metadata.version("pip"))
except metadata.PackageNotFoundError:
    print("PIP metadata is unavailable in this runtime")

try:
    print("requests:", metadata.version("requests"))
except metadata.PackageNotFoundError:
    print("requests is not installed in this runtime")
````

The installation command is shown for reference only; it is not executed:

```python
%pip install requests
```

The terminal equivalent is `python -m pip install requests`. A `requirements.txt` file lists third-party dependencies, not standard-library modules. The Learning Assistant needs no third-party dependency for dates or totals.

**EXPERIMENT:** Classify these as built-in, standard library or third party: `sum`, `datetime`, `requests`.

**REFLECT:** Why is avoiding an unnecessary dependency useful? Fewer dependencies usually mean simpler installation and fewer compatibility concerns.

## 7. Exploring an External Library (3 minutes)

**CONCEPT:** An HTTP request asks a web service for data. `requests` is a popular third-party library for making HTTP requests, while `datetime` is part of Python's standard library. A future application might use:

```python
import requests
response = requests.get("https://example.com")
```

That example is deliberately not executed: this notebook must not depend on network access.

**PREDICT:** Which import needs a separate package installation in a fresh Python installation: `from datetime import date` or `import requests`?

````python
from datetime import date
requirements = {
    "calculate a total from a list": "built-in function: sum",
    "find a square root": "standard-library module: math",
    "calculate a date seven days from now": "standard-library module: datetime",
    "send an HTTP request": "third-party package: requests",
}

for requirement, choice in requirements.items():
    print(requirement, "->", choice)

assert date(2026, 10, 28).isoformat() == "2026-10-28"
````

**REFLECT:** Standard-library code is already available; third-party code may need installation, version selection and documentation checks. Never add a dependency without a reason.
