## 2. Creating Your First Class (3 minutes)

**CONCEPT:** The `class` keyword defines a new type. `pass` is a placeholder for a class with no behaviour yet. Calling the class creates an object. Each call creates a separate instance.

**PREDICT:** What will the type name be? Will `session1 is session2` be `True` or `False`? What about `session3 is session1`?

````python
class EmptyStudySession:
    pass

session1 = EmptyStudySession()
session2 = EmptyStudySession()
session3 = session1

print(type(session1).__name__)
print(session1 is session2)
print(session3 is session1)
assert type(session1).__name__ == "EmptyStudySession"
assert session1 is not session2
assert session3 is session1
````

## 3. Constructors and Instance Attributes (4 minutes)

**CONCEPT:** `__init__()` is the initialisation method called after a new instance is created. `self` refers to the instance currently being initialised. `self.topic` and `self.duration` are instance attributes, so every object can hold different values.

Line by line: `class` defines the type; `__init__()` receives the new values; `self.topic = topic` stores a topic on this object; `self.duration = duration` stores its duration.

**PREDICT:** What four values will the next cell print?

````python
class StudySession:
    def __init__(self, topic, duration):
        self.topic = topic
        self.duration = duration

python_session = StudySession("Python", 45)
maths_session = StudySession("Mathematics", 30)

print(python_session.topic)
print(python_session.duration)
print(maths_session.topic)
print(maths_session.duration)
assert python_session.topic == "Python"
assert maths_session.duration == 30
````

## 4. Understanding `self` (3 minutes)

**CONCEPT:** An instance method is a function defined inside a class. When we call `first.describe()`, Python passes `first` as the instance argument, conventionally named `self`. Therefore `self.topic` can be different for each object.

**PREDICT:** What two descriptions will be returned? They should use the state of the object on which each method is called.

````python
class StudySession:
    def __init__(self, topic, duration):
        self.topic = topic
        self.duration = duration

    def describe(self):
        return f"{self.topic}: {self.duration} minutes"

first = StudySession("Python", 45)
second = StudySession("English", 25)
print(first.describe())
print(second.describe())
assert first.describe() == "Python: 45 minutes"
assert second.describe() == "English: 25 minutes"
````

The corrected method includes `self`, even if it does not use an attribute.

## 5. Instance Methods and Object Behaviour (4 minutes)

**CONCEPT:** Attributes represent an object's data; methods represent operations related to that data. Methods can accept extra parameters and should return values when callers may need to reuse, display or test those values.

**PREDICT:** For a 45-minute session, will `to_hours()` return `0.75`? What will `is_focused()` return for thresholds 30, 45 and 60?

````python
class StudySession:
    def __init__(self, topic, duration):
        self.topic = topic
        self.duration = duration

    def describe(self):
        return f"{self.topic}: {self.duration} minutes"

    def to_hours(self):
        return round(self.duration / 60, 2)

    def is_focused(self, minimum_duration):
        return self.duration >= minimum_duration

session = StudySession("Python", 45)
print(session.to_hours())
print(session.is_focused(30))
print(session.is_focused(45))
print(session.is_focused(60))
assert session.to_hours() == 0.75
assert session.is_focused(45) is True
assert session.is_focused(60) is False
````

## 6. Independent Object State (3 minutes)

**CONCEPT:** Each object has independent instance state. Changing one object's attribute does not automatically change another object's attribute. A class attribute can be shared, which is dangerous for a mutable list of sessions.

**PREDICT:** After changing `first.duration`, what will `second.duration` be?

````python
first = StudySession("Python", 45)
second = StudySession("Mathematics", 30)
first.duration = 60
print(first.duration)
print(second.duration)
assert first.duration == 60
assert second.duration == 30
````

This is an incorrect design because every tracker can refer to the same class list:

```python
class SharedStudyTracker:
    sessions = []
```

Put the list on `self` inside `__init__()` instead. Then every tracker receives its own list.

````python
class IndependentStudyTracker:
    def __init__(self):
        self.sessions = []

tracker_one = IndependentStudyTracker()
tracker_two = IndependentStudyTracker()
tracker_one.sessions.append(StudySession("Python", 45))
print(len(tracker_one.sessions), len(tracker_two.sessions))
assert tracker_one.sessions is not tracker_two.sessions
assert len(tracker_one.sessions) == 1
assert len(tracker_two.sessions) == 0
````

## 7. Working with Collections of Objects (4 minutes)

**CONCEPT:** A list can contain objects. A repeated topic remains a separate study-session object because each activity may have a different duration.

**PREDICT:** What is the total? Which durations will remain after filtering for at least 30 minutes? Repeated `"Python"` topics should remain separate.

````python
sessions = [
    StudySession("Python", 45),
    StudySession("Mathematics", 30),
    StudySession("Python", 60),
    StudySession("English", 25),
]

total = sum([session.duration for session in sessions])
focused = [
    session
    for session in sessions
    if session.is_focused(30)
]
focused_durations = [session.duration for session in focused]
print(total)
print(focused_durations)
assert total == 160
assert focused_durations == [45, 30, 60]
assert len([session for session in sessions if session.topic == "Python"]) == 2
````

## 8. Build a Simple `StudyTracker` (4 minutes)

**CONCEPT:** `StudySession` represents one activity. `StudyTracker` manages multiple activity objects. Storing objects keeps each topic and duration together and lets the tracker call their attributes or methods.

**PREDICT:** What will `total_minutes()` and `get_durations()` return after three sessions?

````python
class StudyTracker:
    def __init__(self):
        self.sessions = []

    def add_session(self, session):
        self.sessions.append(session)

    def total_minutes(self):
        return sum([session.duration for session in self.sessions])

    def get_durations(self):
        return [session.duration for session in self.sessions]

tracker = StudyTracker()
tracker.add_session(StudySession("Python", 45))
tracker.add_session(StudySession("Mathematics", 30))
tracker.add_session(StudySession("Python", 60))
print(tracker.total_minutes())
print(tracker.get_durations())
assert tracker.total_minutes() == 135
assert tracker.get_durations() == [45, 30, 60]
````

**MODIFY:** Add a `session_count()` method returning `len(self.sessions)`. Then call it for the tracker. A complete reference class is shown below:

````python
class StudyTracker:
    def __init__(self):
        self.sessions = []

    def add_session(self, session):
        self.sessions.append(session)

    def total_minutes(self):
        return sum([session.duration for session in self.sessions])

    def get_durations(self):
        return [session.duration for session in self.sessions]

    def session_count(self):
        return len(self.sessions)

reference_tracker = StudyTracker()
reference_tracker.add_session(StudySession("Python", 45))
reference_tracker.add_session(StudySession("Mathematics", 30))
reference_tracker.add_session(StudySession("Python", 60))
assert reference_tracker.session_count() == 3
print("Session count:", reference_tracker.session_count())
````
