## 2. Variables and Naming Conventions (4 minutes)

A variable is a meaningful name referring to a value. Assignment uses one equals sign (`=`): it stores the value on the right under the name on the left. It is not mathematical equality. Use lowercase `snake_case`, such as `weekly_hours`.

````python
student_name = "Alex"
weekly_hours = 4.5
print(f"{student_name} studies {weekly_hours} hours each week.")
````

````python
weekly_hours = weekly_hours + 1.5
student_name = "Alex Morgan"
print(f"Updated: {student_name} now studies {weekly_hours} hours each week.")
````

## 3. Built-in Data Types (4 minutes)

`int` stores whole numbers, `float` decimal numbers, `str` text, and `bool` `True` or `False`. `None` means a value is absent or not set; its type is `NoneType`.

````python
whole_number = 24
decimal_number = 4.5
text_value = "24"
boolean_value = True
missing_value = None

print(repr(whole_number), "->", type(whole_number).__name__)
print(repr(decimal_number), "->", type(decimal_number).__name__)
print(repr(text_value), "->", type(text_value).__name__)
print(repr(boolean_value), "->", type(boolean_value).__name__)
print(repr(missing_value), "->", type(missing_value).__name__)
````

### Predict ? Run

Predict the five type names in the next cell. `24` is an `int`; `"24"` is a `str`, even though the characters look like a number.

````python
prediction_values = ["course", 7, 4.5, False, None]
print(f"{prediction_values[0]!r}: {type(prediction_values[0]).__name__}")
print(f"{prediction_values[1]!r}: {type(prediction_values[1]).__name__}")
print(f"{prediction_values[2]!r}: {type(prediction_values[2]).__name__}")
print(f"{prediction_values[3]!r}: {type(prediction_values[3]).__name__}")
print(f"{prediction_values[4]!r}: {type(prediction_values[4]).__name__}")
````

## 4. Type Conversion and Numerical Operations (5 minutes)

`input()` always returns a string (text), even when a learner enters digits. Convert text explicitly when you need a number:

````python
age_as_number = int("24")
hours_as_number = float("4.5")
number_as_text = str(24)
print(age_as_number, type(age_as_number).__name__)
print(hours_as_number, type(hours_as_number).__name__)
print(number_as_text, type(number_as_text).__name__)
````

````python
weekly_hours = float("4.5")
course_weeks = 7
total_hours = weekly_hours * course_weeks
print(total_hours)
````

````python
print(4 * 3)
print("4" * 3)
````

Numerical multiplication gives `12`, while string repetition gives `"444"`: text and numbers follow different rules.

## 5. Working with Strings (5 minutes)

Strings are sequences of characters. You can join them, measure them, select characters, take slices, and use methods that return transformed strings.

````python
full_name = "  aLeX moRGan  "
clean_name = full_name.strip().title()
print(clean_name)
````

````python
full_name = "Alex Morgan"
print("Characters:", len(full_name))
print("First character:", full_name[0])
print("Last character:", full_name[-1])
print("First four characters:", full_name[:4])
````

### Predict ? Run

Predict the first character, last character and first four characters of `"Alex Morgan"`. Indexing starts at `0`; `[-1]` means the final character; `[:4]` stops before position `4`.

````python
first_name = clean_name.split()[0]
name_parts = clean_name.split()
initials = name_parts[0][0].upper() + name_parts[1][0].upper()
print("First name:", first_name)
print("Initials:", initials)
````

````python
messy_name = "  aLeX moRGan  "
print(messy_name.strip())
print(messy_name.strip().lower())
print(messy_name.strip().upper())
print(messy_name.strip().title())
print(messy_name.strip().split())
````

### Try It

### Reflect

Strings are **immutable**: methods such as `.upper()` return a new string rather than changing the original automatically. Assign the result when you want to keep it.

## 6. Formatting Output with f-Strings (3 minutes)

An f-string puts variables and expressions inside `{}` so a message remains readable.

````python
student_name = "Alex"
weekly_hours = 4.5
course_weeks = 7
print(f"{student_name} plans to study {weekly_hours * course_weeks} hours.")
````

````python
student_name = "Alex"
weekly_hours = 4.5
course_weeks = 7
readable = f"{student_name} plans to study {weekly_hours * course_weeks} hours."
less_readable = student_name + " plans to study " + str(weekly_hours * course_weeks) + " hours."
print(readable)
print(less_readable)
````
