## 2. The `for` Loop: Repeating a Known Number of Times (4 minutes)

**Concept:** Use `for` when the number of repetitions is known. In `for session in range(3):`, `for` begins the loop, `session` is the loop variable, and `range(3)` supplies three values: `0`, `1` and `2`. The indented body runs once for each value.

**Predict:** How many times will each of the two loops below execute? What values will `session` have?

**Run:** The first loop counts from zero. The second uses `range(1, 4)` to display session numbers from 1 through 3.

````python
for session in range(3):
    print("Study session completed")

for session in range(1, 4):
    print(f"Session {session}")
````

**Experiment:** Change `range(3)` to `range(5)`. Predict the new number of messages before you run the cell again.

## 3. Understanding `range()` and Off-by-One Errors (3 minutes)

**Concept:** `range(stop)` starts at 0 and stops before `stop`. `range(start, stop)` starts at `start` and stops before `stop`. `range(start, stop, step)` also sets the amount to count by.

**Predict:** Which numbers will the following ranges display? Remember that each stop value is excluded.

````python
for number in range(4):
    print(number)

for number in range(2, 5):
    print(number)

for number in range(2, 11, 2):
    print(number)
````

**Debugging activity — find the boundary error:** The Learning Assistant should print session numbers 1 to 5, but this loop displays only 1 to 4. Predict its output, then identify how to fix the stop value.

**Run:** This is a genuine logic error, not a syntax error.

````python
for session in range(1, 5):
    print(f"Session {session}")
````

**Corrected solution:** Because the stop value is excluded, use 6 to include session 5. `range(0)` is a boundary case with zero iterations: its body does not run.

**Experiment:** Change the corrected range to `range(1, 6)` or `range(0)` and describe the difference.

**Reflect:** Why are off-by-one errors common in loops?

````python
for session in range(1, 6):
    print(f"Session {session}")

print("The next loop has zero iterations:")
for session in range(0):
    print(f"Session {session}")
````

## 4. Counters and Accumulators (4 minutes)

**Concept:** A **counter** records how many times something happens. An **accumulator** stores a running total. Both must be set before the loop so their values are kept between iterations.

| Session | Duration added | Completed-session counter | Total minutes |
| --- | ---: | ---: | ---: |
| Before loop | — | 0 | 0 |
| 1 | 15 | 1 | 15 |
| 2 | 30 | 2 | 45 |
| 3 | 45 | 3 | 90 |

**Predict:** What total and counter will the Study Time Calculator show after three iterations?

````python
total_minutes = 0
completed_sessions = 0

for session in range(1, 4):
    duration = session * 15
    total_minutes += duration
    completed_sessions += 1

print(f"Completed sessions: {completed_sessions}")
print(f"Total study time: {total_minutes} minutes")
````

**Experiment:** Change the loop to run for five sessions. Predict the new total first.

**Reflect:** What would happen if `total_minutes = 0` were placed inside the loop? Why must the starting value be set before the repeated work?

## 5. The `while` Loop: Repeating Until a Condition Changes (4 minutes)

**Concept:** Use `while` when repetition depends on a condition rather than a known count. Python checks the condition before every iteration. The body updates the state, and the loop ends when the condition becomes false.

**Predict:** How many progress messages will appear? What will the final value of `total_minutes` be?

**Run:** The initial value is 0, the condition checks whether it is less than 60, and each iteration adds 20.

````python
total_minutes = 0

while total_minutes < 60:
    total_minutes += 20
    print(f"Progress: {total_minutes} minutes")
````

**Experiment:** Change the target from 60 to 70. Does the loop stop exactly at the target?

**Reflect:** What would happen if `total_minutes` never changed? A `while` loop needs a state update that can eventually make its condition false.

## 6. `break`: Stopping a Loop Early (3 minutes)

**Concept:** `break` ends the nearest enclosing loop immediately. Here, the learner decides to stop tracking when session 3 is reached.

**Predict:** Which session numbers will be recorded? Will the message for session 3 appear?

**Run:**

````python
for session in range(1, 6):
    if session == 3:
        print("Tracking stopped.")
        break

    print(f"Recording session {session}")
````

**Experiment:** Change the stop condition from session 3 to session 4.

**Reflect:** Why might a programme need to stop before completing all planned iterations?

## 7. `continue`: Skipping an Iteration (3 minutes)

**Concept:** `continue` skips the remaining statements in the current iteration, then the loop moves on to the next value. It does not end the whole loop.

**Predict:** What happens after session 3 is skipped? Will sessions 4 and 5 still be shown?

**Run:**

````python
for session in range(1, 6):
    if session == 3:
        print("Session skipped.")
        continue

    print(f"Session {session} completed.")
````

**Compare:** `break` ends the loop; `continue` skips only the current iteration.

**Experiment:** Replace `continue` with `break` and predict which later sessions disappear.

**Reflect:** Why is it important to distinguish skipping one iteration from ending the loop?

### A loop that may never finish

**Concept — read only; do not run:**

```python

while total_minutes < 60:
    print(total_minutes)
```

The condition remains true because the loop never changes `total_minutes`. This is a potentially infinite loop. Recognise the risk by checking whether the loop body changes something that affects its condition. Test with a small target and verify that the condition eventually becomes false. ## 9. Boundary Testing and Final Reflection (3 minutes)

**Concept:** Engineering Thinking means checking not only the typical case, but also the boundaries. A Study Progress Tracker should behave sensibly for one planned session, several sessions, a skipped session, stopping immediately, stopping later and no completed sessions.

**Predict:** The tracker below stops at `stop_before_session`. Predict its completed-session count and total time for each setting before changing the values. In particular, what happens when it stops before session 1?

````python
planned_sessions = 4
stop_before_session = 3
completed_sessions = 0
total_minutes = 0

for session in range(1, planned_sessions + 1):
    if session == stop_before_session:
        print("Tracking stopped.")
        break

    duration = 20
    completed_sessions += 1
    total_minutes += duration
    print(f"Session {session} recorded: {duration} minutes")

print(f"Completed sessions: {completed_sessions}")
print(f"Total study time: {total_minutes} minutes")
````

**Experiment:** Try these cases one at a time, resetting the values at the top of the code cell each time:

- One planned session: set `planned_sessions = 1` and `stop_before_session = 2`.
- Several planned sessions: leave the starting values unchanged.
- A skipped session: in the loop, skip one chosen session with `continue` rather than `break`.
- Stopping immediately: set `stop_before_session = 1`.
- Stopping after several completed sessions: set `stop_before_session = 4`.
- No completed sessions: set `stop_before_session = 1` and check both totals.

**Engineering Insight:** A programme that works for three normal sessions may still fail when the learner stops immediately. Initialising the counter and total before the loop means both remain valid at zero even when the loop performs no work.
